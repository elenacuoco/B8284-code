## What Python is and what you are running

Python is an **interpreted** language: you write a line, it runs, it answers.
Nothing is compiled first, which is why a notebook can work one cell at a
time and why an error appears only when the line actually runs.

It is also **dynamically typed**: you never declare that something is a number
before using it as one. Python works out what a value is when it meets it.
That is what makes it quick to write and it is why the errors you meet are
usually about types.

Three things are running when you open this page:

| Piece | What it does |
|---|---|
| the **interpreter** | the Python that executes your lines |
| the **kernel** | the process holding your variables between cells |
| the **notebook** | this page, which sends cells to the kernel |

The distinction matters exactly once and then constantly. If the kernel is
restarted, every variable you defined is gone and the cells above have to be
run again — even though the results are still printed on the page. Output on
the screen is not the same as a value in memory.

## Variables: giving a number a name

A variable is a name for a value. The `=` sign does not mean equality here; it
means *store this under that name*.


````
arm_length = 4000        # metres, one LIGO arm
strain     = 1e-21       # dimensionless, a typical wave at the Earth

length_change = strain * arm_length
print(length_change)
````

That is the whole of gravitational wave detection in three lines: an arm of
four kilometres, a wave of strain $10^{-21}$ and a movement of
$4\times10^{-18}$ metres — about a thousandth of the diameter of a proton.

Look closely at what was printed: `3.9999999999999995e-18`, not `4e-18`. That
is not a bug and it is not Python being careless. A computer stores a `float`
in a fixed number of binary digits, so most decimal numbers are held to about
sixteen significant figures and no further. The error is in the sixteenth
digit and it is irrelevant here.

It is worth knowing early for one reason: **never test two floats for exact
equality.** `0.1 + 0.2 == 0.3` is `False` in Python, in C and in every other
language that uses the same arithmetic. Ask whether the difference is small
instead, which is what `np.isclose` does.

Some habits worth having from the first day.

**Write the units in a comment.** Python does not know that `arm_length` is in
metres. You do, for about a week. The comment is for the version of you that
comes back in November.

**Names are for humans.** `arm_length` costs six more keystrokes than `L` and
saves an hour of confusion. Names can contain letters, digits and underscores
and cannot start with a digit.


## Types: what kind of thing a value is

Every value has a **type** and the type decides what you can do with it.


````
print(type(4000))             # a whole number
print(type(1e-21))            # a number with a decimal point
print(type("H1"))             # text
print(type(True))             # true or false
````

Four types carry almost everything in this course:

| Type | What it is | Example from this course |
|---|---|---|
| `int` | a whole number | `4096`, the samples per second |
| `float` | a number with a decimal point | `1e-21`, a strain |
| `str` | text, in quotes | `"H1"`, the name of a detector |
| `bool` | `True` or `False` | whether a detector was observing |

`1e-21` is Python's way of writing $1\times10^{-21}$. You will type a lot of
these. `1e-21` is a `float`; so is `2.5`; so is `4000.0`. Note that `4000` and
`4000.0` are different types, even though they are the same number.

The type matters because operations behave differently:


````
print(3 * 2)          # int times int
print("H" * 3)        # str times int -- repetition, not multiplication
print(7 / 2)          # division always gives a float
print(7 // 2)         # floor division: whole part only
print(7 % 2)          # remainder
````

## Converting between types

Python will not silently turn one type into another, so you do it yourself.
The three conversion functions are named after the types.


````
print(int("4096"), type(int("4096")))       # text to whole number
print(float("1e-21"), type(float("1e-21")))  # text to float
print(str(4096), type(str(4096)))            # number to text

print(int(7.9))          # float to int TRUNCATES, it does not round
print(round(7.9))        # this rounds
````

`int(7.9)` gives `7`. It throws the fractional part away rather than rounding,
which is occasionally what you want and frequently a bug. Use `round` when you
mean rounding.

## When it breaks — and it will

This is the most useful part of the session, so we do it on purpose.

Run the next cell. **It is meant to fail.** Read what comes out before doing
anything else.


````
print(arm_lenght)
````

You should see a red block ending in something like

```
NameError: name 'arm_lenght' is not defined
```

Read a traceback from the **bottom up**. The last line says what went wrong;
the arrow above it says where.

Here the message is exact: there is no variable called `arm_lenght`, because
the name was typed with the letters swapped. Python does not guess. Fix the
spelling in the cell above and run it again.

Three errors will account for most of your first weeks:

| Message | What it means | Usual cause |
|---|---|---|
| `NameError` | that name does not exist | a typo or the cell defining it was never run |
| `TypeError` | wrong kind of thing | doing arithmetic on text |
| `SyntaxError` | Python cannot read the line | a missing bracket or quote |

The second one is worth meeting too. Run this and read the message:


````
samples = "4096"
print(samples * 2)      # this is not an error, but it is not what you meant
print(samples + 2)      # this one raises TypeError
````

The first line printed `40964096`: text repeated, not a number doubled. The
second failed outright, because Python will not add a number to text.

The cure is to convert:


````
samples = int("4096")     # now it is a number
print(samples * 2)
print(type(samples))
````

> **Important — The single habit that matters most**
>
> When a cell fails, read the last line of the error before doing anything else.
> It is a sentence in English and it is almost always accurate.
>
> Beginners scroll past it, change something at random and run again. That is
> the one habit that turns a five second problem into a lost afternoon.


## Your first plot

Plotting is a library job. A **library** is code somebody else wrote; you get
it with `import`.


````
import numpy as np
import matplotlib.pyplot as plt

print("libraries loaded")
````

`import numpy as np` means: load the library called `numpy` and let me refer
to it as `np`. Everybody writes those two lines exactly like that, so copy the
convention rather than inventing your own.

Now a plot. We will draw a simple wave — not real data yet, that is session 4,
but the shape a detector spends its life looking for.


````
t = np.linspace(0, 0.2, 2000)      # 2000 time points between 0 and 0.2 s
f = 100.0                          # hertz
h = 1e-21 * np.sin(2 * np.pi * f * t)

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(t, h)
ax.set_xlabel("time (s)")
ax.set_ylabel("strain")
ax.set_title("A 100 Hz wave with strain amplitude 1e-21")
plt.show()
````

Four lines do the work and the rest is labelling. Get used to labelling.

> **Careful — An axis without a label is not a result**
>
> Every plot in this course must say what is on each axis and in what units. A
> plot without labels cannot be checked by anybody, including you in a week.
>
> This is not a style preference. Half the mistakes in data analysis are caught
> by noticing that an axis says something impossible.

Notice `np.linspace(0, 0.2, 2000)`: it made two thousand evenly spaced numbers
between $0$ and $0.2$. That object is an **array** and it is the single most
important thing in scientific Python. Session 3 is about arrays. For now it is
enough to know that `np.sin` applied to an array gives you the sine of every
element at once, without a loop.


## How to find out what a function does

You will spend this course calling functions other people wrote. Learning to
read their instructions is worth more than learning to write your own.

Put the cursor inside the brackets of a function and press **Shift+Tab** to
see its documentation. Or ask directly:


````
help(np.linspace)
````

The important line is the signature at the top:

```
linspace(start, stop, num=50, endpoint=True, ...)
```

`start` and `stop` must be given. `num` has a default of `50`, so you may
leave it out — or set it **by name**, which is clearer and is what you should
prefer:


````
a = np.linspace(0, 1, 5)              # by position
b = np.linspace(0, 1, num=5)          # by name -- same thing, easier to read
print(a)
print(b)
````
